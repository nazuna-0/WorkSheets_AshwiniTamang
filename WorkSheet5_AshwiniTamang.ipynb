Exercise 5: Cost Function Validation (Test Case)

```python
def train_test_split_scratch(X, Y, test_size=0.2, seed=42):
    # spliting data
    np.random.seed(seed)
    idx = np.arange(len(X))
    np.random.shuffle(idx)

    #no of test samples
    split = int(len(X) * test_size)

    #test indices
    test_idx = idx[:split]
    #training indices
    train_idx = idx[split:]


    return X.iloc[train_idx], X.iloc[test_idx], Y.iloc[train_idx], Y.iloc[test_idx]

X_train, X_test, Y_train, Y_test = train_test_split_scratch(X, Y)
print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("Y_train shape:", Y_train.shape)
print("Y_test shape:", Y_test.shape)
```

```python

# Define the cost function
def cost_function(X, Y, W):
    #no. of samples
    n = len(Y)
    #predicted values
    Y_pred = np.dot(X, W)
    #formual
    cost = (1 / (2 * n)) * np.sum((Y_pred - Y) ** 2)
    return cost

```

**• To - Do - 3:**

Exercise 6: Implementing Gradient Descent from Scratch

Exercise 7: Gradient Descent Verification

```python
def gradient_descent(X, Y, W, alpha, iterations):
    m = len(Y)
    cost_history = []

    for _ in range(iterations):
        Y_pred = np.dot(X, W)
        loss = Y_pred - Y
        dw = (1 / m) * np.dot(X.T, loss)
        W = W - alpha * dw
        cost = cost_function(X, Y, W)
        cost_history.append(cost)

    return W, cost_history
```

**• To - Do - 4:**

Exercise 8: Root Mean Squared Error (RMSE) Function

Exercise 9: Coefficient of Determination (R^2) Function

```python

def rmse(Y, Y_pred):
    return np.sqrt(np.mean((Y - Y_pred) ** 2))
```

```python
def r2(Y, Y_pred):
    mean_y = np.mean(Y)
    ss_tot = np.sum((Y - mean_y) ** 2)
    ss_res = np.sum((Y - Y_pred) ** 2)
    return 1 - (ss_res / ss_tot)
```

**Main Function to Integrate All Steps:**

Exercise 10: The Main Execution Pipeline

xercise 11: Analysis and Findings

```python
def main():

    data = pd.read_csv("student.csv")


    X = data[['Math', 'Reading']].values
    Y = data['Writing'].values

    X_train, X_test, Y_train, Y_test = train_test_split_scratch(X, Y, test_size=0.2, seed=42)


    W = np.zeros(X_train.shape[1])
    alpha = 0.0001
    iterations = 1000


    W_opt, cost_history = gradient_descent(X_train, Y_train, W, alpha, iterations)


    Y_pred = np.dot(X_test, W_opt)


    print("\nFinal Weights:", W_opt)
    print("RMSE on Test Set:", rmse(Y_test, Y_pred))
    print("R2 Score on Test Set:", r2(Y_test, Y_pred))
    print("Cost History (First 10 iterations):", cost_history[:10])


if __name__ == "__main__":
    main()
```
