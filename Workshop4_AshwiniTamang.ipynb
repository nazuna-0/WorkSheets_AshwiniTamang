```python
cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

for c in cols:
    # change all the 0 values and replace with cols mean.
    diab[c] = diab[c].replace(0, diab[c].mean())

diab.isnull().sum()
```

```python
X = diab.drop(columns=['Outcome']).values
y = diab['Outcome'].values

print(X.shape)
print(y.shape)
```

```python
def train_test_split_scratch(X, y, test_size=0.3, seed=42):
    # fixes randomness
    np.random.seed(seed)
    # row of data set
    idx = np.arange(len(X))\
    # randomize the pattern to avoid sequential data selection
    np.random.shuffle(idx)

    split = int(len(X) * test_size)
    test_idx = idx[:split]
    train_idx = idx[split:]                # Train indices

    X_train = X[train_idx]
    X_test = X[test_idx]
    y_train = y[train_idx]
    y_test = y[test_idx]

    return X_train, X_test, y_train, y_test

X_train, X_test, y_train, y_test = train_test_split_scratch(X, y)
```

```python
def euclidean_distance(p1, p2):
    if p1.shape != p2.shape:
        raise ValueError("Dimensions must match")
    return np.sqrt(np.sum((p1 - p2) ** 2))
```

```python
def knn_predict_single(query, X_train, y_train, k=3):
    distances = [euclidean_distance(query, x) for x in X_train]
    nearest = np.argsort(distances)[:k]
    labels = y_train[nearest]
    return np.bincount(labels).argmax()
```

```python
def knn_predict(X_test, X_train, y_train, k=3):
    return np.array([knn_predict_single(x, X_train, y_train, k) for x in X_test])
```

```python
def compute_accuracy(y_true, y_pred):
    return (np.sum(y_true == y_pred) / len(y_true)) * 100
```

```python
pred = knn_predict(X_test, X_train, y_train, k=3)
acc = compute_accuracy(y_test, pred)

print(f"Accuracy (Original Data): {acc:.2f}%")
```
